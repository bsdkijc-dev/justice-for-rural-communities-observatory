# 01 — Authoritative Source Inventory & Pilot Geography

**Project:** Justice for Rural Communities Observatory (JRC Observatory)  
**Approved Pilot Provinces:** Nusa Tenggara Barat (`52`) & Jawa Barat (`32`)  
**Purpose:** Inspect verified Indonesian public sources (`sources/source-register.csv`), check metadata completeness, and review acquisition connector readiness.

```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve() if Path('../sources').exists() else Path('.').resolve()
sources_df = pd.read_csv(ROOT / 'sources' / 'source-register.csv')
with (ROOT / 'src' / 'acquisition' / 'pilot_config.json').open('r', encoding='utf-8') as f:
    pilot_cfg = json.load(f)

print(f'Loaded {len(sources_df)} verified sources.')
print('Approved pilot:', pilot_cfg['pilot_name'], '| Status:', pilot_cfg['approval_status'])
for p in pilot_cfg['provinces']:
    print(f"  - {p['province_code']} {p['province_name']}: {p['geography_profile']}")
```

```python
tier_summary = sources_df.groupby(['publication_tier', 'verification_status']).size().reset_index(name='count')
print(tier_summary.to_string(index=False))

missing_urls = sources_df[~sources_df['exact_url'].str.startswith('https://')]
print(f'Sources missing valid HTTPS URL: {len(missing_urls)}')
```
